```python
# ==========================================
# PLAN DATOS - FASE 0: DÍA 3
# Control de Flujo Defensivo y Tipado Estático
# ==========================================

import pandas as pd
from typing import List, Dict, Any

# 1. FUNCIÓN CON TYPE HINTS DIRIGIDA A PANDAS / INGENIERÍA DE DATOS[cite: 1]
def process_data(df: pd.DataFrame) -> dict:
    """Procesa un DataFrame de compras y retorna métricas de calidad y volumen."""
    if df.empty:
        return {"status": "EMPTY_DATAFRAME", "total_records": 0, "valid_volume_chf": 0.0}
    
    total_records: int = len(df)
    # Filtrado defensivo mediante control de flujo lógico sobre vectores
    valid_df = df[df["amount_chf"] > 0]
    invalid_records: int = total_records - len(valid_df)
    total_volume: float = float(valid_df["amount_chf"].sum())

    return {
        "status": "SUCCESS",
        "total_records": total_records,
        "invalid_records": invalid_records,
        "valid_volume_chf": round(total_volume, 2)
    }

# 2. BUCLE WHILE: Paginación de Ingesta desde una API (Simulación)[cite: 1]
print("--- Simulación de Paginación en API ---")
page: int = 1
max_pages: int = 3

while page <= max_pages:
    print(f"Ingiriendo paquete de datos desde API (Página {page}/{max_pages})...")
    page += 1

# 3. EJECUCIÓN CON UN DATAFRAME DE PRUEBA
data_payload = {
    "order_id": ["ORD-101", "ORD-102", "ORD-103", "ORD-104"],
    "amount_chf": [1500.0, -50.0, 450.0, 3200.0],
    "country": ["CH", "DE", "FR", "CH"]
}

raw_df = pd.DataFrame(data_payload)
metrics_summary: dict = process_data(raw_df)

print(f"\nResultado del Pipeline: {metrics_summary}")
```
