```python
def process_data(df: pd.DataFrame) -> dict:
    if df.empty:
        return {"status": "Empty_DATAFRAME", "total_records": 0, "valid_volume_chf": 0.0}

    total_records: int = len(df)
    valid_df = df[df["amount_chf"] > 0]
    invalid_records: int = total_records - len(valid_df)
    total_volume: float = (valid_df["amount_chf"].sum())

    return{
        "status": "SUCCESS",
        "total_records": total_records,
        "invalid_records": invalid_records,
        "valid_volume_chf": round(total_volume, 2)
    }
```

```python
def leer_monto_archivo(ruta_archivo: str) -> float:
    """Lee y convierte un monto de archivo con control de excepciones."""
    try:
        with open(ruta_archivo, "r", encoding="utf-8") as file:
            contenido = file.read().strip()
            monto = float(contenido)
            return monto
    except FileNotFoundError:
        print(f"⚠️ Error: El archivo '{ruta_archivo}' no existe.")
        return 0.0
    except ValueError:
        print(f"⚠️ Error: El contenido del archivo no se puede convertir a número decimal.")
        return 0.0
    finally:
        print("Fin del intento de lectura de archivo.")

# Prueba de control de excepciones
resultado = leer_monto_archivo("archivo_inexistente.txt")
```
