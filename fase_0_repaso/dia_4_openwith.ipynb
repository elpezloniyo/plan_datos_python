```python
def leer_monto_archivo(ruta_archivo: str) -> float:
    """Lee y convierte un monto de archivo con control de excepciones."""
    try:
        with open(ruta_archivo, "r", encoding="utf-8") as file:
            contenido = file.read().strip()
            monto = float(contenido)
            return monto
    except FileNotFoundError:
        print(f"⚠️ Error: El archivo '{ruta_archivo}' no existe.")
        return 0.0
    except ValueError:
        print(f"⚠️ Error: El contenido del archivo no se puede convertir a número decimal.")
        return 0.0
    finally:
        print("Fin del intento de lectura de archivo.")

# Prueba de control de excepciones
resultado = leer_monto_archivo("archivo_inexistente.txt")
```
