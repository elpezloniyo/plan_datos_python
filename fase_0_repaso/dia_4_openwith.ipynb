```python
logging.basicConfig(
    filename="fase_0_repaso/pipeline.log",
    level = logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    encoding="utf-8"
)
def procesar_registro_pipeline(monto: float) -> None:
    logging.info(f"Iniciando procesamiento de registro con monto:{monto} CHF")
    if monto <= 0:
        logging.warning(f"Monto anómalo detectado ({monto} Chf). Marcado para revisión.")
    else:
        logging.info("Registro procesado con éxito.")

procesar_registro_pipeline(1500.0)
procesar_registro_pipeline(-20.0)
```
