```python
import pandas as pd

# 1. Definir payload
payload = {
    "transaction_id": ["TX-101", "TX-102", "TX-103", "TX-104"],
    "amount_chf": [1500.50, 450.00, 3200.00, 89.90],
    "country": ["CH", "DE", "CH", "AT"],
    "status": ["COMPLETED", "COMPLETED", "PENDING", "COMPLETED"]
}

# 2. Crear DataFrame
df_pandas = pd.DataFrame(payload)

# 3. Imprimir título
print("--- Inspección inicial de memoria en Pandas ---")

# 4. Ejecutar inspección en su propia línea
df_pandas.info(verbose=False)
```

```python
df_pandas["country"] = df_pandas["country"].astype("category")
df_pandas["status"] = df_pandas["status"].astype("category")

print("\n--- Inspección tras optimización (astype) ---")
df_pandas.info(verbose=False)
```

```python
print(df_pandas.columns)
```

```python
df_pandas.to_parquet("transaccions_pandas.parquet", index=False)
print("\n[OK] Archivo 'transactions_pandas.parquet' exportado con éxito")
```
