```python
capitales = {'El salvador': 'San Salvador', 'Honduras': 'Tegucigalpa', 
             "Nicaragua": "Managua", "Costa Rica": "San Jose", 
"Panama": "Panama", "Argentina": "Buenos Aires", 
"Colombia": "Bogota", "Venezuela": "Caracas", 
"España": "Madrid"
             }
```

```python
pais = input(str('Ingrese un país para conocer su capital'))
paismayuscula = pais.capitalize() in capitales
if paismayuscula == True:
    print(capitales[pais.capitalize()])
else:
    print('El país no está en el diccionario')
```
